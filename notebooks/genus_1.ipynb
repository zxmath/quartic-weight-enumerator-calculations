# Genus one

## Fixed-curve counts

The counts $B_d(E)$ in Proposition 3.7 follow from Propositions 3.5–3.6. Write $n=\#E(\mathbf F_q)$ and $e=\#E[2](\mathbf F_q)$.


Split, doubled, and nonsplit divisors number $\binom n2$, $n$, and $(\#E(\mathbf F_{q^2})-n)/2$, respectively, with $\#E(\mathbf F_{q^2})=n(2q+2-n)$.

```python
R.<q,n,e> = PolynomialRing(QQ)
K = R.fraction_field()
n2 = n*(2*q+2-n)
divisors_total = {'split':binomial(n,2),'doubled':n,'nonsplit':(n2-n)/2}
print(table([(kind,K(value).factor()) for kind,value in divisors_total.items()],
           header_row=['divisor type','total']))
```

For one singular point, Proposition 3.5 gives $q-2$, $q-1$, and $q$ centers for split, doubled, and nonsplit fibers, excluding the support and cone vertex. Multiply by the divisor counts.

```python
one_point = {
 'split tacnode':R(divisors_total['split']*(q-2)),
 'ramphoid cusp':R(divisors_total['doubled']*(q-1)),
 'nonsplit tacnode':R(divisors_total['nonsplit']*q),
}
print(table([(kind,value.factor()) for kind,value in one_point.items()],
           header_row=['singularity','series']))
```

For two singular points, subtract equal-class pairs from disjoint pairs. There are $n/e$ classes with $e$ doubled divisors and $n-n/e$ classes with none. A class with $r$ doubled divisors has $(n-r)/2$ split and $q+1-r-(n-r)/2$ nonsplit divisors.

Pairs of the same type are unordered, and different types are distinguished. The four profiles below have nonzero trace shift.

```python
class_types = [(n/e,e),(n-n/e,0)]
equal_class = {label:K.zero() for label in ['two split','split and cusp','nonsplit and cusp','two nonsplit']}
for number_of_classes,doubled in class_types:
    split = (n-doubled)/2
    nonsplit = q+1-doubled-split
    equal_class['two split'] += number_of_classes*binomial(split,2)
    equal_class['split and cusp'] += number_of_classes*split*doubled
    equal_class['nonsplit and cusp'] += number_of_classes*nonsplit*doubled
    equal_class['two nonsplit'] += number_of_classes*binomial(nonsplit,2)
disjoint = {
 'two split':3*binomial(n,4),
 'split and cusp':n*binomial(n-1,2),
 'nonsplit and cusp':n*divisors_total['nonsplit'],
 'two nonsplit':binomial(divisors_total['nonsplit'],2),
}
two_points = {label:R(disjoint[label]-equal_class[label]) for label in disjoint}
print(table([(label,K(disjoint[label]).factor(),equal_class[label].factor(),two_points[label].factor())
            for label in disjoint],
           header_row=['profile','disjoint supports','equal classes to remove','quartic series']))
```

A split node or tacnode contributes $1$ to the trace shift, a nonsplit one contributes $-1$, and a cusp contributes zero. Each of the $n$ degree-four line bundles has $q^3+q^2+q+1-n$ allowed centers. Subtract the $n$ exceptional conic series, then the four nonzero-shift counts to obtain $B_0(E)$.

```python
B = {
 2:two_points['two split'],
 1:two_points['split and cusp']+one_point['split tacnode'],
 -1:two_points['nonsplit and cusp']+one_point['nonsplit tacnode'],
 -2:two_points['two nonsplit'],
}
all_centers = n*(q^3+q^2+q+1-n)
conic_series = n
total = all_centers-conic_series
B[0] = total-sum(B.values())
print(table([(d,B[d].factor()) for d in [2,1,0,-1,-2]],header_row=['d','B_d(E)']))
print('Total =',total.factor())
```

Export the polynomials to `genus_one_fixed.json` for the weighted trace calculation below.


```python
from pathlib import Path
import json
result_dir = Path('../results')
result_dir.mkdir(exist_ok=True)
payload = {'scope':'fixed genus-one series derived from divisor counts','variables':['q','n','e'],'B':{str(d):str(value) for d,value in B.items()}}
(result_dir / 'genus_one_fixed.json').write_text(json.dumps(payload, indent=2) + '\n')
print('Saved', result_dir / 'genus_one_fixed.json')
```

## Weighted trace counts

The elliptic distributions use [Kaplan–Petrow, §3, Lemmas 2–3](https://arxiv.org/abs/1510.03980). Multiply their $1/(q\#\mathrm{Aut}(E,O))$-weighted formulas by $q$ for our normalization.


Cancel translations using $\#\mathrm{Aut}(E)=n\#\mathrm{Aut}(E,O)$ and substitute $n=q+1-t$ in $B_d/n$. The result is linear in $e=\#E[2](\mathbf F_q)$.

```python
from pathlib import Path
import json
input_ring = PolynomialRing(QQ,names=('q','n','e'))
input_q,input_n,input_e = input_ring.gens()
data = json.loads(Path('../results/genus_one_fixed.json').read_text())
fixed_B = {int(d):input_ring(value) for d,value in data['B'].items()}
R.<q,t,e,epsilon,E_count,F_count> = PolynomialRing(QQ)
substitute_trace = input_ring.hom([q,q+1-t,e],R)
after_translations = {d:substitute_trace(input_ring(value/input_n))
                      for d,value in fixed_B.items()}
print(table([(d,after_translations[d]) for d in [2,1,0,-1,-2]],
           header_row=['d','B_d(E) / n after substituting n']))
```

With $\epsilon=\mathbf1_{2\mid t}$, the weighted sums of $1$ and $e$ are $\mathscr E_q(t)$ and $(1+\epsilon)\mathscr E_q(t)+2\mathscr F_q(t)$. In the code, `E_count` and `F_count` denote $\mathscr E_q(t)$ and $\mathscr F_q(t)$.

```python
weighted_B = {}
for d,value in after_translations.items():
    constant = value.subs({e:0})
    linear = value.coefficient({e:1})
    weighted_B[d] = (constant+linear*(1+epsilon))*E_count+2*linear*F_count
print(table([(d,weighted_B[d].coefficient({E_count:1}),weighted_B[d].coefficient({F_count:1}))
            for d in [2,1,0,-1,-2]],
           header_row=['d','coefficient of E_q(t)','coefficient of F_q(t)']))
print('Sum over shifts =',sum(weighted_B.values()).factor())
```

For Lemma 4.2, `h_order(D)` counts primitive quadratic-form classes, with weights $1/3$ at $D=-3$ and $1/2$ at $D=-4$. `hurwitz_kronecker(D)` sums over square divisors. The formulas include ordinary, supersingular, and full rational two-torsion cases.

```python
from collections import defaultdict
from sage.rings.number_field.order import quadratic_order_class_number


def prime_power_data(q0):
    factors = ZZ(q0).factor()
    if len(factors) != 1:
        raise ValueError('q must be a prime power')
    p, v = factors[0]
    if p <= 3:
        raise ValueError("The manuscript assumes characteristic greater than three")
    return ZZ(p), ZZ(v)


def h_order(D):
    D = ZZ(D)
    if D >= 0 or D % 4 not in [0, 1]:
        return QQ(0)
    h = quadratic_order_class_number(D)
    if D == -3:
        return QQ(h) / 3
    if D == -4:
        return QQ(h) / 2
    return QQ(h)


def hurwitz_kronecker(D):
    D = ZZ(D)
    if D >= 0:
        return QQ(0)
    return QQ(pari(-D).qfbhclassno())

def elliptic_weighted_count_formula(q0, trace):
    p, v = prime_power_data(q0)
    trace = ZZ(trace)
    Delta = trace*trace - 4*q0
    if trace*trace < 4*q0 and trace % p != 0:
        return hurwitz_kronecker(Delta) / 2
    if v % 2 == 1 and trace == 0:
        return hurwitz_kronecker(-4*p) / 2
    if v % 2 == 0 and trace == 0:
        return QQ(1 - kronecker_symbol(-4, p)) / 4
    if v % 2 == 0 and trace*trace == q0:
        return QQ(1 - kronecker_symbol(-3, p)) / 6
    if v % 2 == 0 and trace*trace == 4*q0:
        return QQ(p - 1) / 24
    return QQ(0)

def full_two_torsion_weighted_count_formula(q0, trace):
    p, v = prime_power_data(q0)
    trace = ZZ(trace)
    Delta = trace*trace - 4*q0
    if trace*trace < 4*q0 and trace % p != 0 and (trace - q0 - 1) % 4 == 0:
        return hurwitz_kronecker(Delta // 4) / 2
    if q0 % 4 == 3 and trace == 0:
        return h_order(-p) / 2
    if trace*trace == 4*q0:
        return elliptic_weighted_count_formula(q0, trace)
    return QQ(0)


def torsion_weighted_count_formula(q0, trace):
    E = elliptic_weighted_count_formula(q0, trace)
    F = full_two_torsion_weighted_count_formula(q0, trace)
    return (1 + ZZ(trace % 2 == 0))*E + 2*F
```

Evaluate at $q=5,7$. Other prime powers of characteristic greater than three can be added to `field_sizes`.

```python
def calculate_field(q0):
    q0 = ZZ(q0)
    prime_power_data(q0)
    bound = ZZ(4*q0).isqrt()
    elliptic,full,coefficients = {},{},{}
    for trace in range(-bound,bound+1):
        E = elliptic_weighted_count_formula(q0,trace)
        F = full_two_torsion_weighted_count_formula(q0,trace)
        elliptic[trace],full[trace] = E,F
        for d,value in weighted_B.items():
            coefficients[(trace,d)] = QQ(value.subs({q:q0,t:trace,epsilon:ZZ(trace%2==0),
                                                    E_count:E,F_count:F,e:0}))
    return elliptic,full,coefficients

field_sizes = [5,7]
weighted_results = {ZZ(q0):calculate_field(q0) for q0 in field_sizes}
for q0,(elliptic,full,coefficients) in weighted_results.items():
    print('q =',q0)
    print(table([(trace,elliptic[trace],full[trace])
                +tuple(coefficients[(trace,d)] for d in [2,1,0,-1,-2])
                for trace in sorted(elliptic)],
               header_row=['t','E_q(t)','F_q(t)','B_2','B_1','B_0','B_-1','B_-2']))
```
