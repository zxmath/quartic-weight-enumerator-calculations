# Genus zero

Source: Jonas Bergström, *Cohomology of moduli spaces of curves of genus three via point counts*, [§12.3, Table 6 and §13.4, Lemma 13.5](https://arxiv.org/pdf/math/0611815). The three-singularity counts are imported from the odd-characteristic column of Table 6. The two-singularity counts are calculated from Lemma 13.5.

Counts are normalized as $\overline m_\tau=m_\tau(\mathbf P^1)/\#\mathrm{PGL}_2(\mathbf F_q)$. Labels `c`, `sp`, `ns` denote a cusp, split node, nonsplit node. Labels `tc`, `tsp`, `tns` denote a ramphoid cusp, split tacnode, nonsplit tacnode. A suffix `(i)` denotes a Frobenius orbit of $i$ singularities, with the indicated type over its field of definition.

Only rational singularities contribute to the trace shift. Split nodes and tacnodes contribute $1$, nonsplit nodes and tacnodes $-1$, and cusps $0$.

## Two and three singular points

### Three singular points

The 22 rows of Table 6 are stored as (profile, normalized count, trace shift). Table 6 evaluates Lemma 12.13, equation (5), excluding the three-fiber configurations in a degree-two pencil that give conic images.

```python
R.<q> = PolynomialRing(QQ)
K = R.fraction_field()

three_singularities = [
 ("3c", 1/6, 0),
 ("2c+sp", (q-3)/4, 1),
 ("2c+ns", (q-1)/4, -1),
 ("c+2sp", (q-3)^2/8, 2),
 ("c+sp+ns", (q-1)^2/4, 0),
 ("c+2ns", (q-1)^2/8, -2),
 ("3sp", (q-3)*(q^2-7*q+11)/48, 3),
 ("2sp+ns", (q-1)*(q^2-3*q+1)/16, 1),
 ("sp+2ns", (q^3-2*q^2-1)/16, -1),
 ("3ns", (q-1)*(q^2-3*q+1)/48, -3),
 ("c+c(2)", 1/2, 0),
 ("sp+c(2)", (q-1)/4, 1),
 ("ns+c(2)", (q-3)/4, -1),
 ("c+sp(2)", (q^2-2*q-1)/4, 0),
 ("c+ns(2)", (q^2-1)/4, 0),
 ("sp+sp(2)", (q-1)*(q^2-q-1)/8, 1),
 ("ns+sp(2)", (q-3)*(q^2-q-3)/8, -1),
 ("sp+ns(2)", (q+1)*(q^2-q+1)/8, 1),
 ("ns+ns(2)", (q-1)*(q^2-q-1)/8, -1),
 ("c(3)", 1/3, 0),
 ("sp(3)", (q^3-q^2-q-3)/6, 0),
 ("ns(3)", (q-1)*(q^2+1)/6, 0),
]
print(table(three_singularities, header_row=["profile", "normalized count", "shift"]))
```

### Two singular points

In Lemma 13.5, $\lambda$ and $\mu$ describe the Frobenius orbits in the supports of the delta-two and delta-one fibers. The normalized count is
\[
\frac{|\mathbf P^1(\lambda)|\,|X(\mu)|\,(q-\lambda_1)}
     {z(\lambda)z(\mu)\,\#\mathrm{PGL}_2(\mathbf F_q)},
\qquad X=\mathbf P^1\setminus\{\text{the chosen }\lambda\text{-set}\},
\]
where $z(\lambda)=\prod_i i^{\lambda_i}\lambda_i!$ and $\lambda_1$ counts rational points in the first support.

Counting unordered supports directly incorporates the factors $z(\lambda)$ and $z(\mu)$. A doubled fiber has one rational support point, a split fiber two, and a nonsplit fiber one degree-two closed point. After choosing disjoint supports, multiply by $q-\lambda_1$ and divide by $\#\mathrm{PGL}_2$. The different delta invariants distinguish the two fibers.

For a split tacnode and a split node this gives $\binom{q+1}{2}\binom{q-1}{2}(q-2)/\bigl(q(q^2-1)\bigr)=(q-2)^2/4$.

```python
# Available closed points on P1 of degrees one and two.
available = {1: q+1, 2: (q^2-q)/2}
support = {
    "cusp": {1:1, 2:0},
    "split": {1:2, 2:0},
    "nonsplit": {1:0, 2:1},
}
shift = {"cusp":0, "split":1, "nonsplit":-1}

def unordered_supports(kind, remaining):
    if kind == "cusp":
        return remaining[1]
    if kind == "split":
        return remaining[1]*(remaining[1]-1)/2
    return remaining[2]

two_singularities = []
formula_steps = []
for first, prefix in [("cusp","tc"),("split","tsp"),("nonsplit","tns")]:
    first_choices = unordered_supports(first, available)
    remaining = {i:available[i]-support[first][i] for i in [1,2]}
    extensions = q-support[first][1]
    for second, suffix in [("cusp","c"),("split","sp"),("nonsplit","ns")]:
        second_choices = unordered_supports(second, remaining)
        count = K(first_choices*second_choices*extensions/(q*(q^2-1)))
        label = prefix+"+"+suffix
        two_singularities.append((label,count,shift[first]+shift[second]))
        formula_steps.append((label,first_choices,second_choices,extensions,count.factor()))

print(table(formula_steps,header_row=['profile','first support','second support',
                                     'extensions','normalized count']))
```

### Counts by trace shift

```python
def group_by_shift(rows):
    return {d:K(sum(value for _,value,shift in rows if shift==d)) for d in range(-3,4)}
three = group_by_shift(three_singularities)
two = group_by_shift(two_singularities)
print(table([(d,(three[d].factor() if three[d] else 0),(two[d].factor() if two[d] else 0)) for d in range(3,-4,-1)],
           header_row=['shift','three singularities','two singularities']))
print('Totals =',sum(three.values()).factor(),sum(two.values()).factor())
```

`../results/bergstrom_aggregation.json` saves the grouped counts for the calculation below.

```python
from pathlib import Path
import json
result_dir = Path('../results')
result_dir.mkdir(exist_ok=True)
payload = {'scope':'normalized counts from Bergstrom Table 6 and Lemma 13.5','three':{str(d):str(value) for d,value in three.items()},'two':{str(d):str(value) for d,value in two.items()}}
(result_dir / 'bergstrom_aggregation.json').write_text(json.dumps(payload, indent=2) + '\n')
print('Saved', result_dir / 'bergstrom_aggregation.json')
```

## One singular point

In Proposition 3.11, $C_d(q)$ counts series on the fixed $\mathbf P^1$, without division by $\#\mathrm{PGL}_2$ or the scalar factor $q-1$. The two- and three-singularity counts are calculated above. The one-singularity counts use Lemma 3.8 and Propositions 3.9 and 3.10.

Input: `../results/bergstrom_aggregation.json`.

A degree-$f$ closed point of multiplicity $e$ contributes $q^{(e-1)f}(q^f-1)$ units to $H^0(\mathcal O_M)$. For a degree-three fiber, the extension count is $\#H^0(\mathcal O_M)^\times/(q-1)$. For a degree-two fiber, the pencil count is the same quotient, multiplied by $q-1$ for the nonzero extension parameter.

```python
from pathlib import Path
import json
R.<q> = PolynomialRing(QQ)
K = R.fraction_field()
G = q*(q^2-1)
data = json.loads(Path('../results/bergstrom_aggregation.json').read_text())
three = {int(d):K(value) for d,value in data['three'].items()}
two = {int(d):K(value) for d,value in data['two'].items()}
def units(local_factors):
    return prod(q^((multiplicity-1)*degree)*(q^degree-1)
                for degree,multiplicity in local_factors)
```

Entries of `local_factors` are (closed-point degree, multiplicity). The trace shift is the number of rational support points minus one. Divide the fiber count times the series per fiber by $\#\mathrm{PGL}_2$ to match the normalization above.

```python
fibers = [
 ('D4 1+1+1',(q+1)*q*(q-1)/6,[(1,1),(1,1),(1,1)]),
 ('D4 1+2',(q+1)*(q^2-q)/2,[(1,1),(2,1)]),
 ('D4 3',(q^3-q)/3,[(3,1)]),
 ('D5',q*(q+1),[(1,2),(1,1)]),
 ('E6',q+1,[(1,3)]),
 ('A5 split',(q+1)*q/2,[(1,1),(1,1)]),
 ('A5 nonsplit',(q^2-q)/2,[(2,1)]),
 ('A6',q+1,[(1,2)]),
]
one_rows = []
for label,choices,local_factors in fibers:
    degree = sum(f*e for f,e in local_factors)
    per_fiber = K(units(local_factors)/(q-1) if degree==3 else units(local_factors))
    shift = sum(f==1 for f,e in local_factors)-1
    normalized = K(choices*per_fiber/G)
    one_rows.append((label,choices,per_fiber,shift,normalized))
one = {d:sum(value for _,_,_,shift,value in one_rows if shift==d) for d in range(-3,4)}
print(table([(label,choices,per_fiber.factor(),shift,value.factor())
            for label,choices,per_fiber,shift,value in one_rows],
           header_row=['profile','fibers','series per fiber','shift','normalized count']))
```

## Counts by trace shift

### Nonzero shifts

$C_d(q)=\#\mathrm{PGL}_2(\mathbf F_q)$ times the sum of the normalized one-, two-, and three-singularity counts.

```python
C = {d:R(G*(three[d]+two[d]+one[d])) for d in [-3,-2,-1,1,2,3]}
print(table([(d,(three[d].factor() if three[d] else 0),(two[d].factor() if two[d] else 0),(K(one[d]).factor() if K(one[d]) else 0),C[d].factor())
            for d in [3,2,1,-1,-2,-3]],
           header_row=['d','three singularities','two singularities','one singularity','C_d(q)']))
```

### Total and zero shift

From $\#\mathrm{Gr}(3,5)(\mathbf F_q)$, subtract the subspaces with base divisor of degree two or exactly one. The nonbirational base point free series are symmetric squares of unique base point free pencils in $H^0(\mathcal O(2))$. Subtract these conic series to obtain the total birational count, then subtract the six nonzero shifts to obtain $C_0(q)$.

```python
from sage.combinat.q_analogues import q_binomial

all_subspaces = q_binomial(5,3,q)
base_degree_two = 1+q+q^2
base_degree_one = (q+1)*(q_binomial(4,3,q)-(q+1))
base_point_free = all_subspaces-base_degree_two-base_degree_one
conic_series = q_binomial(3,2,q)-(q+1)
total = base_point_free-conic_series
C[0] = total-sum(C.values())
print(table([('all subspaces',all_subspaces),('base degree two',base_degree_two),
            ('base degree one',base_degree_one),('conic series',conic_series),
            ('birational series',total.factor())],header_row=['step','count']))
print(table([(d,C[d].factor()) for d in range(3,-4,-1)],header_row=['d','C_d(q)']))
```
