# Reducible quartics
Projective counts $R_z(q)$ for the line and conic configurations in Section 2. `weight_enumerator.ipynb` computes the line–cubic contribution from the cited affine-cubic formulas and the elliptic input in `genus_1.ipynb`.

Exact polynomial arithmetic, with $N=q^2+q+1$.

```python
from collections import defaultdict
R.<q> = PolynomialRing(QQ)
K = R.fraction_field()
N = q^2+q+1


def print_rows(rows):
    print(table([(label,z,K(value).factor()) for label,z,value in rows],
               header_row=['configuration','rational zeros','projective count']))
```

Degree-$d$ line orbits are counted over $\mathbf F_{q^d}$, after removing lines defined over proper subfields and dividing by $d$. Smooth rational conics form one $\mathrm{PGL}_3$ orbit with stabilizer $\mathrm{PGL}_2$.

```python
line_orbits = {1:N}
for d in [2,3,4]:
    line_orbits[d] = R((q^(2*d)+q^d+1
                        -sum(e*line_orbits[e] for e in divisors(d) if e<d))/d)
d2,d3,d4 = [line_orbits[d] for d in [2,3,4]]
smooth_conics = R(q^3*(q^3-1)*(q^2-1)/(q*(q^2-1)))
print(table([(d,line_orbits[d]) for d in line_orbits],
           header_row=['orbit degree','line orbits']))
print('Smooth conics =',smooth_conics.factor())
```

A rational pencil contains $q+1$ rational lines and $(q^2-q)/2$ conjugate pairs. Four rational lines in general position are counted by subtracting the concurrent and exactly-three-concurrent cases from $\binom N4$. Multiplicities distinguish the repeated factors.

```python
# Four geometric lines.  Each tuple is (description, rational zero count,
# projective quartic equations).
four_lines = [
 ("L^4", q+1, N),
 ("L^3M", 2*q+1, N*(N-1)),
 ("L^2M^2", 2*q+1, N*(N-1)/2),
 ("(L L^sigma)^2", 1, N*(q^2-q)/2),
 ("L^2 M R, two intersection points", 3*q, N*binomial(q+1,2)*q^2),
 ("L^2 M R, concurrent", 3*q+1, 3*N*binomial(q+1,3)),
 ("L^2 M M^sigma, off", q+2, N*(q^2-q)*q^2/2),
 ("L^2 M M^sigma, on", q+1, N*(q^2-q)*(q+1)/2),
 ("four rational concurrent", 4*q+1, N*binomial(q+1,4)),
 ("three rational concurrent", 4*q-1, N*binomial(q+1,3)*q^2),
 ("four rational general", 4*q-2, binomial(N,4)-N*binomial(q+1,4)-N*binomial(q+1,3)*q^2),
 ("two rational and conjugate pair concurrent", 2*q+1,
  N*(q^2-q)*binomial(q+1,2)/2),
 ("one rational through conjugate intersection", 2*q+1,
  N*(q^2-q)*(q+1)*q^2/2),
 ("two rational and conjugate pair off", 2*q+2,
  N*(q^2-q)*binomial(q^2,2)/2),
 ("rational plus concurrent degree-three orbit, off", q+2,
  (q^3-q)*N*q^2/3),
 ("rational plus concurrent degree-three orbit, on", q+1,
  N*(q^3-q)*(q+1)/3),
 ("rational plus nonconcurrent degree-three orbit", q+1,
  N*(d3-N*(q^3-q)/3)),
 ("two degree-two orbits, same point", 1, N*binomial((q^2-q)/2,2)),
 ("two degree-two orbits, distinct points", 2,
  binomial(N,2)*((q^2-q)/2)^2),
 ("degree-four orbit concurrent", 1, N*(q^4-q^2)/4),
 ("degree-four orbit nonconcurrent", 0,
  (q^8-q^2-N*(q^4-q^2))/4),
]

print_rows(four_lines)
```

A smooth conic has $q+1$ tangents, $\binom{q+1}{2}$ secants, and $N-(q+1)-\binom{q+1}{2}$ external lines. They meet it in one, two, and zero rational points, respectively. Multiply each line count by the number of smooth conics.

```python
tangent = q+1
secant = binomial(q+1,2)
external = N-tangent-secant
double_line_conic = [(label,2*q+2-r,count*smooth_conics)
                     for label,r,count in [('tangent',1,tangent),
                                           ('secant',2,secant),
                                           ('external',0,external)]]
print_rows(double_line_conic)
```

For a fixed conic, a tangent and a secant meet on it in $(q+1)q$ ways, and two distinct secants in $(q+1)\binom q2$ ways. Two secants meeting off the conic correspond to three pairings of four conic points. The remaining counts are products or unordered pairs of the line counts.

```python
two_lines_conic = [
 ('two tangents',3*q,binomial(tangent,2)*smooth_conics),
 ('tangent and secant, on',3*q,(q+1)*q*smooth_conics),
 ('tangent and secant, off',3*q-1,(tangent*secant-(q+1)*q)*smooth_conics),
 ('tangent and external',3*q+1,tangent*external*smooth_conics),
 ('two secants, on',3*q-1,(q+1)*binomial(q,2)*smooth_conics),
 ('two secants, off',3*q-2,3*binomial(q+1,4)*smooth_conics),
 ('secant and external',3*q,secant*external*smooth_conics),
 ('two external lines',3*q+2,binomial(external,2)*smooth_conics),
]
print_rows(two_lines_conic)
```

Let $a_i$ count unordered sets of $i$ rational points with no three collinear, and $c_i(q)$ the smooth conics through a fixed such set. Double counting gives $a_i c_i(q)=(q^5-q^2)\binom{q+1}{i}$.

```python
a = [1,N,binomial(N,2),N*(N-1)*q^2/6,N*(N-1)*q^2*(q-1)^2/24]
c = [R(smooth_conics*binomial(q+1,i)/a[i]) for i in range(5)]
print(table([(i,K(a[i]).factor(),c[i].factor()) for i in range(5)],
           header_row=['i','point sets a_i','smooth conics c_i(q)']))
```

The marked counts for rational conic pairs and conjugate pairs are $a_i\binom{c_i(q)}2$ and $a_i(c_i(q^2)-c_i(q))/2$. If $N_r$ counts pairs with exactly $r$ rational intersection points, each equals $\sum_{r=i}^4\binom ri N_r$. Invert this relation from $i=4$ down to zero.

```python
marked_rational = [K(a[i]*binomial(c[i],2)) for i in range(5)]
marked_conjugate = [K(a[i]*(c[i](q=q^2)-c[i])/2) for i in range(5)]

def remove_extra_intersections(marked):
    counts = {}
    for i in reversed(range(5)):
        counts[i] = R(marked[i]-sum(binomial(r,i)*counts[r] for r in range(i+1,5)))
    return counts

rational_pairs = remove_extra_intersections(marked_rational)
conjugate_pairs = remove_extra_intersections(marked_conjugate)
print(table([(i,marked_rational[i].factor(),marked_conjugate[i].factor()) for i in range(5)],
           header_row=['marked points','rational pairs through them','conjugate pairs through them']))
print(table([(r,rational_pairs[r].factor(),conjugate_pairs[r].factor()) for r in range(4,-1,-1)],
           header_row=['exact intersection count r','rational pairs N_r','conjugate pairs N_r']))
```

A conjugate line pair meets at one rational point. For a fixed conic, this point has $q+1$ choices on the conic and $q^2$ choices off it, with $(q^2-q)/2$ line pairs through each point. Include double conics and collect the counts by rational zeros.

```python
conjugate_lines_conic = [
 ('intersection on conic',q+1,(q+1)*(q^2-q)*smooth_conics/2),
 ('intersection off conic',q+2,q^2*(q^2-q)*smooth_conics/2),
]
all_rows = (four_lines+double_line_conic+two_lines_conic+conjugate_lines_conic
            +[('double conic',q+1,smooth_conics)]
            +[('rational conic pair',2*q+2-r,value) for r,value in rational_pairs.items()]
            +[('conjugate conic pair',r,value) for r,value in conjugate_pairs.items()])
grouped = defaultdict(R.zero)
for label,z,count in all_rows:
    grouped[R(z)] += R(count)
print(table([(z,grouped[z].factor()) for z in sorted(grouped,key=lambda z:z(q=101))],
           header_row=['z','R_z(q)']))
```

Export at $q=5,7$, combining symbolic zero counts that coincide after specialization. `weight_enumerator.ipynb` applies the scalar factor $q-1$.

```python
from pathlib import Path
import json
result_dir = Path('../results')
result_dir.mkdir(exist_ok=True)
payload = {'scope':'projective line/conic quartics from Section 2 incidence calculations',
 'fields':{str(q0):{str(z):str(sum(value(q=q0) for exponent,value in grouped.items()
                                if exponent(q=q0)==z))
                   for z in sorted(set(ZZ(exponent(q=q0)) for exponent in grouped))}
           for q0 in [5,7]}}
(result_dir / 'reducible_incidence.json').write_text(json.dumps(payload, indent=2) + '\n')
print('Saved', result_dir / 'reducible_incidence.json')
```
